### zeros() ones() randn()
zeros()创建一个全0张量  
ones()创建一个全1张量  
randn()从某个特定的概率分布中随机采样来得到张量中每个元素的值。每个元素都是从标准正态分布（均值为0，方差为1）中随机抽取的。

```python
print(torch.zeros(2,3,4))
print(torch.ones(2,3,4))
print(torch.randn(3,4))
```

```python
print(torch.tensor([[1,2,3],[4,5,6],[2,5,6]]))
```

### 多个连结 torch.cat()
叠起来形成一个更大的张量，提供张量列表，并给出沿哪个轴连结

```python
X = torch.arange(12,dtype=torch.float32).reshape((3,4))
Y = torch.tensor([[2.0,1,4,3],[1,2,3,4],[4,3,2,1]])
print(torch.cat((X,Y),dim=0))
print(torch.cat((X,Y),dim=1))
```

```python
print(X==Y) #比较两个张量每个是否相等，返回一个布尔值张量
print(X.sum())    #对所有元素进行求和，会形成一个单元素张量tensor(66.)
```

## 2.1.3 广播机制
在某些情况下，我们可以对两个形状不同但兼容的张量进行算术运算。这种机制被称为广播（broadcasting）。

```python
a=torch.arange(3).reshape((3,1))
b=torch.arange(2).reshape((1,2))
print(a)
print(b)
```

```python
print(a+b)  #3*2
```

### 2.1.5 节省内存
某些操作导致为新结果分配内存.  
> 不可取：  
> 1.不必要，数据量大；  
> 2.不断更新，其它引用仍会指向旧内存地址，某些代码会无意识中引用旧的参数
解决方法：切片表示法X[:],或X += Y

```python
before = id(Y)
Y = Y + X
print(id(Y)==before) #False
```

```python
Z=torch.zeros_like(Y)
print('id(Z):',id(Z))
Z[:] = Y + X
print('id(Z):',id(Z)) 
Z += X
print('id(Z):',id(Z))
```
