## 基础术语
### 标量
由只有一个元素的张量标示

```python
import torch
x = torch.tensor(3.0)
y = torch.tensor(4.0)
print(x + y, x * y)
```

### 向量
向量可以被视为标量值组成的列表。

```python
x=torch.arange(4)
print(x) #tensor([0, 1, 2, 3])向量的维度4
print(len(x)) #向量的长度4
print(x.shape)
```

> 明确一下:向量或轴的维度被用来表示向量或轴的长度，即向量或轴的元素数量。然而，张量的维度用来表示张量具有的轴数。在这个意义上，张量的某个轴的维数就是这个轴的长度

### 矩阵

### 张量

```python

A = torch.arange(20,dtype=torch.float32).reshape(5,4)
B = A.clone()   # 通过分配新内存，将A的一个副本分配给B
print(A,'\n',B)
```

```python
print(A*B)
#将张量乘以或加上一个标量不会改变张量的形状，其中张量的每个元素都将与标量相加或相乘。
X = torch.arange(24).reshape(2,3,4)
a = 2
print(a + X)
```

### 2.3.6 降维 axis
默认情况下，调用求和函数会沿所有的轴降低张量的维度，使它变为一个标量。  
我们还可以指定张量沿哪一 个轴来通过求和降低维度。

```python
A_sum_axis0 = A.sum(axis=0)
print(A_sum_axis0) #求和所有行的元素
A_sum_axis1 = A.sum(axis=1)
print(A_sum_axis1) #求和所有列的元素
print(A.sum(axis=[0,1])) #沿行和列对矩阵求和=求和所有元素

#mean同样
```

```python
#在调用函数来计算总和或均值时保持轴数不变
print(A.sum(axis=1,keepdims=True)) #保持轴数不变

#沿某个轴计算A元素的累积总和，不会沿 任何轴降低输入张量的维度。
print(A.cumsum(axis=1))
```

## 2.3.7-2.3.11 矩阵运算
### 点积（内积）
相同位置的按元素乘积的和.  
例：w*x  
将两个向量规范化得到单位长度后，点积表示它们夹角的余弦。

```python
x = torch.arange(4,dtype=torch.float32)
y = torch.ones(4,dtype=torch.float32)
print(torch.dot(x,y))
```

### 向量积
我们可以把一个矩阵A ∈ Rm×n乘法看作一个从Rn到Rm向量的转换。

```python
print(torch.mv(A,x))
```

### 矩阵乘法
可以将矩阵‐矩阵乘法AB看作简单地执行m次矩阵‐向量积

```python
B = torch.ones(4, 3)
print(torch.mm(A, B))
```

## 2.3.10 范数
表示一个向量有多大  
> 性质如下：
> 1.如果我们按常数因子α缩放向量的所有元素，其范数也会按相同常数因子的绝对值缩放；
> f(x + y) ≤ f(x) + f(y)（三角不等式）
> 非负

```python
#欧几里得距离（L2范数）
u=torch.tensor([3.0,-4.0])
print(torch.norm(u))

#曼哈顿距离（L1范数）
#表示为向量元素的绝对值之和
print(torch.abs(u).sum())
print(torch.norm(u,p=1))

#Frobenius范数
#矩阵元素的平方和的平方根
print(torch.norm(torch.ones((4,9))))

#默认 = 所有元素平方相加再开方；要改别的范数，才写 p 参数。
```
