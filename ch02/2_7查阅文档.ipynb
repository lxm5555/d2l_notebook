## Jupyter 查看帮助的三种方法

### 1. `help(list)` —— 输出区打印文档

- 最基础的 Python 内置方式
- 文档直接打印在输出区
- 任何 Python 环境（含普通终端、.py 文件）都能用

### 2. `list?` —— 弹窗显示文档

- Jupyter 特有语法（普通 .py 文件里无效）
- 效果与 help() 几乎相同，但在单独的面板/窗口中显示，方便边看边写
- **适用于：函数、类、方法、模块、变量**（变量会显示类型 + 当前值 + 该类型的文档）
- 忘记参数怎么传时最快，且显示的是当前安装版本的文档

### 3. `list??` —— 显示源代码

- 在 `?` 基础上额外显示 Python 源码实现
- 对库函数和自己写的函数最有用
- 局限：C 语言实现的内置对象（如 list、torch.Tensor）没有 Python 源码可看

### 补充：Tab 补全

- 输入 `对象.` 后按 Tab → 列出该对象的所有可用方法
- 只想快速浏览"能调什么"时，比 `?` 更快

### 对比表

| 方法 | 显示内容 | 环境要求 | 适用对象 |
|------|---------|---------|---------|
| `help(list)` | 文档 | 任何 Python 环境 | 函数/类/模块 |
| `list?` | 文档（弹窗） | 仅 notebook | 函数/类/模块/**变量** |
| `list??` | 文档 + 源码 | 仅 notebook | Python 实现的函数/类 |
| `x.` + Tab | 方法列表 | 仅 notebook | 任何对象 |
