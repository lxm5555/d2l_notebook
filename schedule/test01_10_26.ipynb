# 《动手学深度学习》学习计划（4–5 个月方案）

> 节奏：每天 2–3 小时，边学边跑代码，按周推进
> 起点：今天学完 2.6，下周进入第 3 章

## 总预期

| 目标 | 时间 | 说明 |
|------|------|------|
| 读懂概念、跟完代码 | 3 个月 | 快，但留存率低 |
| **跑完主要实验 + 真正理解（推荐）**&#8203; | **4–5 个月** | 当前学习风格适合这个 |
| 精读推导 + 复现 + 笔记 | 6 个月+ | 适合科班/方向明确的人 |

## 周计划表

### 第 1 周：收尾预备知识 + 线性回归
- 2.6 概率、2.7 查文档（很快）
- 3.1–3.3 线性回归（从零实现 + 简洁实现 + softmax）
- 本章大量使用自动求导，是检验 2.5 学没透的最好机会

### 第 2 周：softmax 回归 + MLP 基础
- 3.6–3.7 softmax
- 4.1–4.3 感知机、MLP 从零实现、简洁实现

### 第 3 周：MLP 进阶 + 模型选择
- 4.4–4.10（权重衰减、dropout、数值稳定性、Kaggle 房价可选）
- 第 5 章深度学习计算（模型构造、参数管理、GPU——本地没 GPU，这段快读）

### 第 4 周：卷积神经网络入门
- 6.1–6.6：从二维卷积到 LeNet
- 第一个 CNN，跑 Fashion-MNIST，本地 CPU 没问题

### 第 5–6 周：现代 CNN（第 7 章）⭐ 分水岭
- AlexNet / VGG / NiN / GoogLeNet / ResNet / 批量归一化
- 从这周起实验开始慢：ResNet 本地可能一个 epoch 半小时
- **开始用 Colab 跑训练，本地读代码**

### 第 7–8 周：RNN（第 8 章）
- 序列模型、语言模型、GRU/LSTM、深层 RNN
- 文本预处理 + 代码量大，两周

### 第 9–10 周：现代 RNN + 注意力（第 9–10 章）
- 编码器-解码器、seq2seq、注意力、Transformer
- **Transformer 是重中之重，多留时间**

### 第 11–12 周：优化算法 + 计算性能（第 11–12 章）
- SGD/Adam、学习率调度
- 并行、多 GPU：没硬件，**只读懂概念，实验全跳**

### 第 13–14 周：计算机视觉（第 13 章，选学）
- 微调、目标检测、语义分割
- **全部在 Colab/Kaggle 跑**

### 第 15–16 周：NLP（第 14–15 章，选学）
- word2vec、BERT、微调
- 同样上云跑

## 两个重要调整

### 1. 李航的安排
- 前 4 周（D2L 基础和 MLP 学完）专攻 D2L
- 从第 5 周起，每周 1–2 天给李航
- 他书里的 CNN/RNN/Transformer 章节和 D2L 互相印证，正好同步
- 老师有明确进度要求的话，以老师为准

### 2. 每周留 1 天不学新内容
- 重跑本周代码（合上书打一遍）
- 整理笔记 / 清理没搞懂的问题
- 缓冲——计划拖了就靠这天追

## 现实提醒

- **第 7 章（第 5–6 周）是最容易弃坑的地方**：代码量突然变大 + 本地跑不动。提前注册好 Colab/Kaggle，无缝切换

- **衡量标准不是"读到第几章"，而是"能不能合上书写出来"**&#8203;：宁可 4.5 个月扎实学完，不要 3 个月囫囵吞
